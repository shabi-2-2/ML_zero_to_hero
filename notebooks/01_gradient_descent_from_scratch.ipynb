# Gradient descent from scratch

A small, dependency-free exercise in fitting a line. The gradient is derived
explicitly so each step can be inspected before moving on to larger models.

```python
xs = [1.0, 2.0, 3.0, 4.0]
ys = [3.0, 5.0, 7.0, 9.0]

weight = 0.0
bias = 0.0
learning_rate = 0.01

for step in range(2000):
    predictions = [weight * x + bias for x in xs]
    errors = [prediction - target for prediction, target in zip(predictions, ys)]
    loss = sum(error * error for error in errors) / len(errors)

    d_weight = sum(2 * error * x for error, x in zip(errors, xs)) / len(xs)
    d_bias = sum(2 * error for error in errors) / len(errors)
    weight -= learning_rate * d_weight
    bias -= learning_rate * d_bias

print(f"weight={weight:.3f}, bias={bias:.3f}, loss={loss:.6f}")
```

```python
assert abs(weight - 2.0) < 0.01
assert abs(bias - 1.0) < 0.01
assert loss < 1e-4
print(f"prediction for x=5: {weight * 5 + bias:.3f}")
```
